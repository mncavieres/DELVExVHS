# RV vs Magellanic Coordinate

```python
fig, ax = plt.subplots(1,1, figsize=(10, 7), sharex=True)  # 4 panels now
ax.scatter(nested_results['L_stream'],nested_results['hel_vel'], s= 15, c=10**nested_results['logage_p84']/1e9, cmap='viridis', alpha=1)
ax.set_xlabel('L_stream (deg)')
ax.set_ylabel('Heliocentric Velocity (km/s)')
cbar = plt.colorbar(ax.collections[0], ax=ax)
cbar.set_label('Age (Gyr)')
plt.show()
```

```python
# mass vs age
fig, ax = plt.subplots(1,1, figsize=(10, 7), sharex=True) 
#ax.scatter(10**nested_results['logage_p50']/1e9, nested_results['mass_p50'], s= 15, c=nested_results['logg'], cmap='plasma', alpha=1)
ax.errorbar(10**nested_results['logage_p50']/1e9, nested_results['mass_p50'],
            xerr=[(10**nested_results['logage_p50'] - 10**nested_results['logage_p16'])/1e9,
                  (10**nested_results['logage_p84'] - 10**nested_results['logage_p50'])/1e9],
            yerr=[nested_results['mass_p50'] - nested_results['mass_p16'],
                  nested_results['mass_p84'] - nested_results['mass_p50']],
            fmt='o', c='C0', alpha=0.7, ecolor='lightgray', elinewidth=2, capsize=4)
ax.vlines(1, ymin=0, ymax=5, colors='red', linestyles='dashed', label='1 Gyr')
ax.vlines(2, ymin=0, ymax=5, colors='orange', linestyles='dashed', label='2 Gyr')
ax.set_xlabel('Age (Gyr)')
ax.set_ylabel('Mass (Msun)')
ax.legend()
cbar = plt.colorbar(ax.collections[0], ax=ax)
cbar.set_label('log(g)')
plt.show()
```
