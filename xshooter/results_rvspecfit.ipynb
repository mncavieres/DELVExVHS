```python
import pickle
import numpy as np
import matplotlib.pyplot as plt
import numpy as np
import holoviews as hv
from astropy.io import fits
hv.extension('bokeh')

with open("/Users/mncavieres/Documents/2024-1/Delve/xshooter/Spec/Star01/res_vacuum_default_full.pkl", "rb") as f:
    vacuum_res = pickle.load(f)


# 3) Load your coadded spectrum from a FITS file
#    (replace "coadd_spectrum.fits" with your actual filepath)
with fits.open('/Users/mncavieres/Documents/2024-1/Delve/xshooter/Spec/Star01/ADP.2025-07-13T14_06_10.099.fits') as hdul:
    # Assume that the primary HDU, or a specific extension, contains a table 
    # with columns "WAVELENGTH" and "FLUX". Adjust if your structure is different.
    data = hdul[1].data
    wavelength = data['WAVE']    # in Angstrom (or whatever units)
    flux       = data['FLUX_REDUCED']          # in erg/s/cm²/Å (or whatever units)

lim_mask = (wavelength> 1200) & (wavelength < 2478)
wavelength = wavelength[lim_mask]
flux = flux[lim_mask]
#espec = espec[lim_mask]

#badmask = (wavelength < 4500) & (wavelength > 3600)#wavelength < 4500 # only fit the first chip


curves = []

# 4) Wrap the arrays into a HoloViews Curve
spectrum_curve = hv.Curve(
    (wavelength*10, flux),
    kdims=['Wavelength (Å)'],
    vdims=['Flux']
).opts(
    title='1D Coadded Spectrum',
    xlabel='Wavelength (Å)',
    ylabel='Flux',
    line_width=1.5,
    # tools=['hover', 'pan', 'wheel_zoom', 'box_zoom', 'reset'],
    # active_tools=['wheel_zoom'],
    color='black', width=1000, height=500
)
# plot the fit
fit_curve = hv.Curve(
    (wavelength*10, vacuum_res['yfit'][0]),
    kdims=['Wavelength (Å)'],
    vdims=['Flux']
).opts(
    title='1D Coadded Spectrum',
    xlabel='Wavelength (Å)',
    ylabel='Flux',
    line_width=1.5,
    # tools=['hover', 'pan', 'wheel_zoom', 'box_zoom', 'reset'],
    # active_tools=['wheel_zoom'],
    color='red', width=1000, height=500
)

curves.extend([spectrum_curve
               , fit_curve])

# 5) Display the interactive plot
#spectrum_curve
# Overlay all curves into one interactive plot.
overlay = hv.Overlay(curves).opts(
    xlabel='Wavelength [$\\AA$]',
    ylabel=r'Flux [$10^{-17}$ erg/s/cm$^2$/{$\\AA$}]',
    legend_position='right'
)

overlay
```
