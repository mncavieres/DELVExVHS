```python
import numpy as np
import matplotlib.pyplot as plt
import astropy.units as u
from astropy.modeling import models, fitting
from specutils import Spectrum1D

# Assuming `spec` is your continuum-normalized Spectrum1D object
# with spec.spectral_axis and spec.flux populated.

def multi_gaussian_fit(spec, line_list, amp_guess=-0.1*u.dimensionless_unscaled,
                       stddev_guess=1.0*u.AA):
    """
    Fit multiple Gaussian absorption features simultaneously to a normalized spectrum.

    Parameters
    ----------
    spec : Spectrum1D
        Continuum-normalized spectrum.
    line_list : list of astropy.units.Quantity
        List of line center guesses in Angstroms.
    amp_guess : astropy.units.Quantity
        Initial guess for Gaussian amplitudes (negative for absorption).
    stddev_guess : astropy.units.Quantity
        Initial guess for Gaussian widths.

    Returns
    -------
    best_model : CompoundModel
        Best-fit multi-Gaussian model.
    fit : LevMarLSQFitter
        Fitter instance after fitting.
    """
    # Extract data
    x = spec.spectral_axis.to(u.AA).value
    y = spec.flux.value

    # Build initial compound model
    models_list = []
    for i, cen in enumerate(line_list):
        g = models.Gaussian1D(amplitude=amp_guess.value,
                              mean=cen.to(u.AA).value,
                              stddev=stddev_guess.to(u.AA).value)
        # Give each model a unique prefix for parameter naming
        g.prefix = f'g{i}_'
        models_list.append(g)

    compound = models_list[0]
    for g in models_list[1:]:
        compound += g

    # Fit
    fitter = fitting.LevMarLSQFitter()
    best_model = fitter(compound, x, y)
    return best_model, fitter
```

```python
import numpy as np
import matplotlib.pyplot as plt
import astropy.units as u
from astropy.modeling import models, fitting
from specutils import Spectrum1D

# If your continuum-normalized spectrum has a baseline around 1,
# you can either subtract 1 before fitting, or include a constant continuum term.

def multi_gaussian_fit(spec,
                       line_list,
                       cont_level=1.0,
                       amp_guess=-0.1,
                       stddev_guess=1.0,
                       bounds=None):
    """
    Fit multiple Gaussian absorption features simultaneously to a normalized spectrum,
    including a constant continuum level.

    Parameters
    ----------
    spec : Spectrum1D
        Continuum-normalized spectrum with baseline around cont_level.
    line_list : list of astropy.units.Quantity
        List of line center guesses in Angstroms.
    cont_level : float
        Approximate continuum level (e.g., 1.0 if normalized to unity).
    amp_guess : float
        Initial guess for Gaussian amplitudes (negative for absorption).
    stddev_guess : float
        Initial guess for Gaussian widths in Angstroms.
    bounds : dict or None
        Optional bounds for parameters, e.g. {{'g0_amplitude': (-1,0)}}

    Returns
    -------
    best_model : CompoundModel
        Best-fit continuum+multi-Gaussian model.
    fitter : LevMarLSQFitter
        Fitter instance.
    """
    # Extract data
    x = spec.spectral_axis.to(u.AA).value
    y = spec.flux.value

    # Subtract continuum if not including a constant term
    # y_fit = y - cont_level
    # Or include a Const1D term below to fit the continuum simultaneously

    # Build continuum + absorption Gaussians
    const = models.Const1D(amplitude=cont_level)
    components = [const]

    for i, cen in enumerate(line_list):
        g = models.Gaussian1D(amplitude=amp_guess,
                              mean=cen.to(u.AA).value,
                              stddev=stddev_guess)
        components.append(g)

    compound = components[0]
    for comp in components[1:]:
        compound += comp

    # Set up fitter
    fitter = fitting.LevMarLSQFitter()

    # Optionally set bounds (requires Astropy >=4.3)
    if bounds:
        for par, (low, high) in bounds.items():
            setattr(compound, par + '_bounds', (low, high))

    # Fit to the data
    best_model = fitter(compound, x, y)
    return best_model, fitter




line_list = [
    3755.0 * u.AA,
    3770.0 * u.AA,
    3800.0 * u.AA,
    3835.384 * u.AA,
    3889.049 * u.AA,
    3933.66 * u.AA,
    3970.076 * u.AA,
    4101.74 * u.AA,
    4340.47 * u.AA,
]

# You can specify bounds if desired:
# bounds = {{'g0_amplitude': (-1, 0), 'g0_stddev': (0.1, 5.0), ...}}
bounds = None

best_model, fitter = multi_gaussian_fit(
    normalized_spec,
    line_list,
    cont_level=1.0,
    amp_guess=-0.1,
    stddev_guess=6.0,
    bounds=bounds
)

# Evaluate the model (continuum + absorption)
x = normalized_spec.spectral_axis.to(u.AA).value
y = normalized_spec.flux.value
y_fit = best_model(x)

# Plotting
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 4), sharey=True)

# Original
ax1.plot(x, y, lw=1)
ax1.set_title('Normalized Spectrum')
ax1.set_xlabel('Wavelength (Å)')
ax1.set_ylabel('Flux')

# Fit Overlay
ax2.plot(x, y, label='Data', lw=1)
ax2.plot(x, y_fit, label='Model', lw=1.5)
ax2.set_title('Multi-Gaussian Fit')
ax2.set_xlabel('Wavelength (Å)')
ax2.legend()

# Residuals
ax3.plot(x, y - y_fit, lw=1)
ax3.axhline(0, color='gray', linestyle='--')
ax3.set_title('Residuals')
ax3.set_xlabel('Wavelength (Å)')

plt.tight_layout()
plt.savefig(os.path.join(plot_dir, 'multi_gaussian_fit.pdf'))
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import astropy.units as u
from astropy.modeling import models, fitting
from specutils import Spectrum1D
from astropy.table import Table

# If your continuum-normalized spectrum has a baseline around 1,
# you can either subtract 1 before fitting, or include a constant continuum term.

def multi_gaussian_fit(spec,
                       line_list,
                       cont_level=1.0,
                       amp_guess=-0.1,
                       stddev_guess=1.0,
                       bounds=None):
    """
    Fit multiple Gaussian absorption features simultaneously to a normalized spectrum,
    including a constant continuum level, and return line parameters.

    Returns:
        best_model: CompoundModel
        fitter: LevMarLSQFitter
        line_table: astropy.table.Table
            Columns: line_guess, center_fit, fwhm, equivalent_width
    """
    # Extract data
    x = spec.spectral_axis.to(u.AA).value
    y = spec.flux.value

    # Build continuum (Const1D) and Gaussian models without prefixes
    const = models.Const1D(amplitude=cont_level)
    components = [const]

    for cen in line_list:
        g = models.Gaussian1D(amplitude=amp_guess,
                              mean=cen.to(u.AA).value,
                              stddev=stddev_guess)
        components.append(g)

    # Combine into a single compound model
    compound = components[0]
    for comp in components[1:]:
        compound += comp

    # Initialize fitter
    fitter = fitting.LevMarLSQFitter()

    # Apply bounds if provided
    if bounds:
        for par, (low, high) in bounds.items():
            setattr(compound, f"{par}_bounds", (low, high))

    # Fit the model
    best_model = fitter(compound, x, y)

    # Extract fitted parameters array and names
    params = best_model.parameters  # numpy array of parameter values
    # First parameter is continuum amplitude
    cont_fit = params[0]

    # Compute line parameters for each Gaussian
    line_results = []
    n_lines = len(line_list)
    for i, cen in enumerate(line_list):
        idx = 1 + 3 * i
        amp = params[idx]
        mean = params[idx + 1]
        sigma = params[idx + 2]
        fwhm = 2 * np.sqrt(2 * np.log(2)) * sigma
        ew = np.sqrt(2 * np.pi) * (-amp) * sigma / cont_fit
        line_results.append((cen.value, mean, fwhm, ew))

    # Create Astropy table of results
    line_table = Table(rows=line_results,
                       names=('line_guess', 'center_fit', 'fwhm', 'equivalent_width'))
    line_table['line_guess'].unit = u.AA
    line_table['center_fit'].unit = u.AA
    line_table['fwhm'].unit = u.AA
    line_table['equivalent_width'].unit = u.AA

    return best_model, fitter, line_table



line_list = [
    #3755.0 * u.AA,
    3770.0 * u.AA,
    3800.0 * u.AA,
    3835.384 * u.AA,
    3889.049 * u.AA,
    3933.664 * u.AA,
    3970.076 * u.AA,
    4101.74 * u.AA,
    4340.47 * u.AA,
]

best_model, fitter, results = multi_gaussian_fit(
    normalized_spec,
    line_list,
    cont_level=1.0,
    amp_guess=-0.1,
    stddev_guess=6.0,
    bounds=None
)

# Evaluate and plot
x = normalized_spec.spectral_axis.to(u.AA).value
y = normalized_spec.flux.value
y_fit = best_model(x)

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 4), sharey=True)
ax1.plot(x, y, lw=1)
ax1.set_title('Normalized Spectrum')
ax1.set_xlabel('Wavelength (Å)')
ax1.set_ylabel('Flux')

ax2.plot(x, y, label='Data', lw=1)
ax2.plot(x, y_fit, label='Model', lw=1.5)
ax2.set_title('Multi-Gaussian Fit')
ax2.set_xlabel('Wavelength (Å)')
ax2.legend()

ax3.plot(x, y - y_fit, lw=1)
ax3.axhline(0, linestyle='--')
ax3.set_title('Residuals')
ax3.set_xlabel('Wavelength (Å)')

plt.tight_layout()
plt.savefig(os.path.join(plot_dir, 'multi_gaussian_fit_with_results.pdf'))
plt.show()

# Display line parameters
print(results)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import astropy.units as u
from astropy.modeling import models, fitting
from specutils import Spectrum1D
from astropy.table import Table

# If your continuum-normalized spectrum has a baseline around 1,
# you can either subtract 1 before fitting, or include a constant continuum term.

def multi_gaussian_fit(spec,
                       line_list,
                       cont_level=1.0,
                       amp_guess=-0.1,
                       stddev_guess=1.0,
                       bounds=None):
    """
    Fit multiple Gaussian absorption features simultaneously to a normalized spectrum,
    including a constant continuum level, and return line parameters.

    Returns:
        best_model: CompoundModel
        fitter: LevMarLSQFitter
        line_table: astropy.table.Table
            Columns: line_guess, center_fit, fwhm, equivalent_width
    """
    # Extract data
    x = spec.spectral_axis.to(u.AA).value
    y = spec.flux.value

    # Build continuum (Const1D) and Gaussian models without prefixes
    const = models.Const1D(amplitude=cont_level)
    components = [const]

    for cen in line_list:
        g = models.Gaussian1D(amplitude=amp_guess,
                              mean=cen.to(u.AA).value,
                              stddev=stddev_guess)
        components.append(g)

    # Combine into a single compound model
    compound = components[0]
    for comp in components[1:]:
        compound += comp

    # Initialize fitter
    fitter = fitting.LevMarLSQFitter()

    # Apply bounds if provided
    if bounds:
        for par, (low, high) in bounds.items():
            setattr(compound, f"{par}_bounds", (low, high))

    # Fit the model
    best_model = fitter(compound, x, y)

    # Extract fitted parameters array and names
    params = best_model.parameters  # numpy array of parameter values
    # First parameter is continuum amplitude
    cont_fit = params[0]

    # Compute line parameters for each Gaussian
    line_results = []
    n_lines = len(line_list)
    for i, cen in enumerate(line_list):
        idx = 1 + 3 * i
        amp = params[idx]
        mean = params[idx + 1]
        sigma = params[idx + 2]
        fwhm = 2 * np.sqrt(2 * np.log(2)) * sigma
        ew = np.sqrt(2 * np.pi) * (-amp) * sigma / cont_fit
        line_results.append((cen.value, mean, fwhm, ew))

    # Create Astropy table of results
    line_table = Table(rows=line_results,
                       names=('line_guess', 'center_fit', 'fwhm', 'equivalent_width'))
    line_table['line_guess'].unit = u.AA
    line_table['center_fit'].unit = u.AA
    line_table['fwhm'].unit = u.AA
    line_table['equivalent_width'].unit = u.AA

    return best_model, fitter, line_table

line_list = [float(x)*u.nm.to(u.AA)*u.AA for x in line_list_df_in_range['obs_wl_vac(nm)'].values]

best_model, fitter, results = multi_gaussian_fit(
    normalized_spec,
    line_list,
    cont_level=1.0,
    amp_guess=-0.1,
    stddev_guess=6.0,
    bounds=None
)

# Evaluate and plot
x = normalized_spec.spectral_axis.to(u.AA).value
y = normalized_spec.flux.value
y_fit = best_model(x)

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 4), sharey=True)
ax1.plot(x, y, lw=1)
ax1.set_title('Normalized Spectrum')
ax1.set_xlabel('Wavelength (Å)')
ax1.set_ylabel('Flux')

ax2.plot(x, y, label='Data', lw=1)
ax2.plot(x, y_fit, label='Model', lw=1.5)
ax2.set_title('Multi-Gaussian Fit')
ax2.set_xlabel('Wavelength (Å)')
ax2.legend()

ax3.plot(x, y - y_fit, lw=1)
ax3.axhline(0, linestyle='--')
ax3.set_title('Residuals')
ax3.set_xlabel('Wavelength (Å)')

plt.tight_layout()
plt.savefig(os.path.join(plot_dir, 'multi_gaussian_fit_with_results.pdf'))
plt.show()

# Display line parameters
print(results)
```

```python
# add barycentric correction
from astropy.io import fits
import numpy as np
import astropy.coordinates as coord
from astropy.coordinates import UnitSphericalRepresentation, CartesianRepresentation
from astropy.coordinates import SkyCoord, solar_system, EarthLocation, ICRS
from astropy.time import Time
from astropy import units as u

def velcorr(time, skycoord, location=None):
  """Barycentric velocity correction.
  code from: https://gist.github.com/StuartLittlefair/5aaf476c5d7b52d20aa9544cfaa936a1

  Uses the ephemeris set with  ``astropy.coordinates.solar_system_ephemeris.set`` for corrections.
  For more information see `~astropy.coordinates.solar_system_ephemeris`.

  Parameters
  ----------
  time : `~astropy.time.Time`
    The time of observation.
  skycoord: `~astropy.coordinates.SkyCoord`
    The sky location to calculate the correction for.
  location: `~astropy.coordinates.EarthLocation`, optional
    The location of the observatory to calculate the correction for.
    If no location is given, the ``location`` attribute of the Time
    object is used

  Returns
  -------
  vel_corr : `~astropy.units.Quantity`
    The velocity correction to convert to Barycentric velocities. Should be added to the original
    velocity.
  """

  # ensure sky location is ICRS compatible
  if not skycoord.is_transformable_to(ICRS()):
    raise ValueError("Given skycoord is not transformable to the ICRS")

  ep, ev = solar_system.get_body_barycentric_posvel('earth', time) # ICRS position and velocity of Earth's geocenter
  op, ov = location.get_gcrs_posvel(time) # GCRS position and velocity of observatory
  # ICRS and GCRS are axes-aligned. Can add the velocities
  velocity = ev + ov # relies on PR5434 being merged

  # get unit ICRS vector in direction of SkyCoord
  sc_cartesian = skycoord.icrs.represent_as(UnitSphericalRepresentation).represent_as(CartesianRepresentation)
  return sc_cartesian.dot(velocity).to(u.km/u.s) # similarly requires PR5434

# read the time of observation from the FITS header
with fits.open('/Users/mncavieres/Documents/2025-1/B576/V4/coadd/b576_hv.fits') as hdul:
    header = hdul[0].header
    # print the header to find the date of observation
    #print(header)
    time_obs = Time(header['MJD'], format='mjd', scale='utc')  # MJD from the header

location = EarthLocation.of_site('gemini_north')#of_site('Mauna Kea')

# define skycoord of the object
coord = SkyCoord(ra=header['RA'], dec=header['DEC'], unit=(u.deg, u.deg), frame='icrs')

correction_to_rv = velcorr(time_obs, coord, location=location)

print(f'The radial velocity is: {np.mean(radial_velocities)*u.km/u.s + correction_to_rv} +-  {np.std(radial_velocities)}')   
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import astropy.units as u
from astropy.modeling import Fittable1DModel, Parameter, fitting, models
from astropy.table import Table
from scipy.special import gamma
from specutils import Spectrum1D

class Seric1D(Fittable1DModel):
    """
    Generalized‐Gaussian (a.k.a. Sersic‐like) profile:
      y = amplitude * exp( - |x - x_0|**c / b )
    """
    amplitude = Parameter(default=-0.1)
    x_0       = Parameter(default=0.0)
    b         = Parameter(default=36.0)   # width measure
    c         = Parameter(default=2.0)    # shape index

    @staticmethod
    def evaluate(x, amplitude, x_0, b, c):
        return amplitude * np.exp(- (np.abs(x - x_0)**c) / b)


def multi_seric_fit(spec,
                    line_list,
                    cont_level=1.0,
                    amp_guess=-0.1,
                    b_guess=36.0,
                    c_guess=2.0,
                    bounds=None):
    """
    Fit multiple Seric absorption features + constant continuum
    to a normalized Spectrum1D.  Returns the best model, fitter,
    and a table of (a, x0, b, c, FWHM, EW) for each line.
    """
    # extract wavelength (Å) and flux
    x = spec.spectral_axis.to(u.AA).value
    y = spec.flux.value

    # continuum
    const = models.Const1D(amplitude=cont_level)
    components = [const]

    # one Seric per line
    for cen in line_list:
        s = Seric1D(amplitude=amp_guess,
                    x_0=cen.to(u.AA).value,
                    b=b_guess,
                    c=c_guess)
        components.append(s)

    # build compound model
    compound = components[0]
    for comp in components[1:]:
        compound += comp

    # attach any bounds dict like {'amplitude_2':(-1,0), 'c_2':(1,5), ...}
    if bounds:
        for parname, (low, high) in bounds.items():
            setattr(compound, parname + '_bounds', (low, high))

    # fit
    fitter = fitting.LevMarLSQFitter()
    best_model = fitter(compound, x, y)

    # continuum level
    cont_fit = best_model.parameters[0]

    # extract per-line results
    results = []
    for i, cen in enumerate(line_list):
        # parameter ordering: [cont_amp,
        #   amp_1, x0_1, b_1, c_1,
        #   amp_2, x0_2, b_2, c_2, …]
        idx = 1 + 4*i
        amp_fit  = best_model.parameters[idx]
        x0_fit   = best_model.parameters[idx+1]
        b_fit    = best_model.parameters[idx+2]
        c_fit    = best_model.parameters[idx+3]

        # analytic FWHM and EW
        fwhm = 2 * (b_fit * np.log(2))**(1.0/c_fit)
        ew   = (-amp_fit/cont_fit) * 2 * (b_fit**(1.0/c_fit) / c_fit) * gamma(1.0/c_fit)

        results.append((cen.value, x0_fit, b_fit, c_fit, fwhm, ew))

    # table
    line_table = Table(rows=results,
                       names=('line_guess', 'center_fit', 'b', 'c', 'fwhm', 'equivalent_width'))
    line_table['line_guess'].unit       = u.AA
    line_table['center_fit'].unit       = u.AA
    # b has units Å^c (so we leave it unitless here)
    line_table['b'].unit                = u.dimensionless_unscaled
    line_table['c'].unit                = u.dimensionless_unscaled
    line_table['fwhm'].unit             = u.AA
    line_table['equivalent_width'].unit = u.AA

    return best_model, fitter, line_table


# --- example usage ---

line_list = [
    3755.0 * u.AA,
    3770.0 * u.AA,
    3800.0 * u.AA,
    3835.384 * u.AA,
    3889.049 * u.AA,
    3933.66 * u.AA,
    3970.076 * u.AA,
    4101.74 * u.AA,
    4340.47 * u.AA,
]

best_model, fitter, results = multi_seric_fit(
    normalized_spec,
    line_list,
    cont_level=1.0,
    amp_guess=-0.3,
    b_guess=36.0,
    c_guess=2.0,
    bounds=None
)

# Evaluate & plot
x = normalized_spec.spectral_axis.to(u.AA).value
y = normalized_spec.flux.value
y_fit = best_model(x)

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 4), sharey=True)
ax1.plot(x, y, lw=1)
ax1.set(title='Normalized Spectrum', xlabel='Wavelength (Å)', ylabel='Flux')

ax2.plot(x, y, label='Data', lw=1)
ax2.plot(x, y_fit, label='Seric Model', lw=1.5)
ax2.set(title='Multi-Seric Fit', xlabel='Wavelength (Å)')
ax2.legend()

ax3.plot(x, y - y_fit, lw=1)
ax3.axhline(0, linestyle='--')
ax3.set(title='Residuals', xlabel='Wavelength (Å)')

plt.tight_layout()
plt.savefig(os.path.join(plot_dir, 'multi_seric_fit.pdf'))
plt.show()

# Display line parameters
print(results)
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import astropy.units as u
from astropy.modeling import Fittable1DModel, Parameter, models, fitting
from astropy.table import Table
from scipy.special import gamma
from specutils import Spectrum1D

class Seric1D(Fittable1DModel):
    """
    Generalized‐Gaussian (Seric‐like) profile:
      y(x) = amplitude * exp( - |x - x_0|**c / b )
    """
    amplitude = Parameter(default=-0.1, bounds=(-1.0, 0.0))
    x_0       = Parameter(default=0.0)
    b         = Parameter(default=36.0, bounds=(1e-6, None))   # enforce b>0
    c         = Parameter(default=2.0, bounds=(0.3, 10.0))     # reasonable shape

    @staticmethod
    def evaluate(x, amplitude, x_0, b, c):
        return amplitude * np.exp(- (np.abs(x - x_0)**c) / b)


def multi_seric_fit(spec,
                    line_list,
                    cont_level=1.0,
                    amp_guess=-0.1,
                    b_guess=36.0,
                    c_guess=2.0):
    """
    Fit multiple Seric absorption lines + fixed continuum to a normalized Spectrum1D.
    Returns:
        best_model : the fitted CompoundModel
        fitter     : the LevMarLSQFitter instance
        line_table : astropy.table.Table with columns
                     line_guess, center_fit, b, c, fwhm, equivalent_width
    """
    # 1) extract and clean data
    x = spec.spectral_axis.to(u.AA).value
    y = spec.flux.value
    finite = np.isfinite(x) & np.isfinite(y)
    x, y = x[finite], y[finite]

    # 2) continuum term (fixed at cont_level)
    const = models.Const1D(amplitude=cont_level)
    const.amplitude.fixed = True
    components = [const]

    # 3) prepare per-line initial guesses
    n_lines = len(line_list)
    if np.isscalar(amp_guess):
        amp_vals = [amp_guess] * n_lines
    else:
        amp_vals = amp_guess
    if np.isscalar(b_guess):
        b_vals = [b_guess] * n_lines
    else:
        b_vals = b_guess
    if np.isscalar(c_guess):
        c_vals = [c_guess] * n_lines
    else:
        c_vals = c_guess

    # 4) build Seric components
    for i, cen in enumerate(line_list):
        s = Seric1D(
            amplitude=amp_vals[i],
            x_0=cen.to(u.AA).value,
            b=b_vals[i],
            c=c_vals[i]
        )
        components.append(s)

    # 5) combine into one CompoundModel
    compound = components[0]
    for comp in components[1:]:
        compound += comp

    # 6) fit with Levenberg–Marquardt
    fitter = fitting.LevMarLSQFitter()
    best_model = fitter(compound, x, y)

    # 7) extract continuum fit (fixed)
    cont_fit = cont_level

    # 8) compute per-line parameters
    results = []
    for i, cen in enumerate(line_list):
        idx = 1 + 4*i
        amp_f = best_model.parameters[idx]
        x0_f  = best_model.parameters[idx+1]
        b_f   = best_model.parameters[idx+2]
        c_f   = best_model.parameters[idx+3]

        # analytic FWHM and EW
        fwhm = 2 * (b_f * np.log(2))**(1.0 / c_f)
        ew   = (-amp_f / cont_fit) * 2 * (b_f**(1.0 / c_f) / c_f) * gamma(1.0 / c_f)

        results.append((cen.value, x0_f, b_f, c_f, fwhm, ew))

    # 9) assemble results table
    line_table = Table(rows=results,
                       names=('line_guess', 'center_fit', 'b', 'c', 'fwhm', 'equivalent_width'))
    line_table['line_guess'].unit       = u.AA
    line_table['center_fit'].unit       = u.AA
    line_table['b'].unit                = u.dimensionless_unscaled
    line_table['c'].unit                = u.dimensionless_unscaled
    line_table['fwhm'].unit             = u.AA
    line_table['equivalent_width'].unit = u.AA

    return best_model, fitter, line_table




# list of hydrogen line centers (Å)
line_list = [
    3755.0 * u.AA,
    3770.0 * u.AA,
    3800.0 * u.AA,
    3835.384 * u.AA,
    3889.049 * u.AA,
    #3933.66 * u.AA,
    3970.076 * u.AA,
    4101.74 * u.AA,
    4340.47 * u.AA,
]

# better guessing??????
x = normalized_spec.spectral_axis.to(u.AA).value
y = normalized_spec.flux.value
amp_guesses = []
b_guesses   = []
for cen in line_list:
    m = (x > cen.value - 5) & (x < cen.value + 5)
    core = y[m]
    amp_guesses.append(np.min(core) - 1.0)  # depth relative to continuum
    # crude width guess: half‐width at half‐min squared
    halfmin = (1.0 + np.min(core))/2.0
    idxs = np.where(core < halfmin)[0]
    if len(idxs) >= 2:
        dx = x[m][idxs[-1]] - x[m][idxs[0]]
        b_guesses.append((dx/2)**2)
    else:
        b_guesses.append(36.0)

# Perform the fit
best_model, fitter, results = multi_seric_fit(
    normalized_spec,
    line_list,
    cont_level=1.0,
    amp_guess=amp_guesses,
    b_guess=b_guesses,
    c_guess=[2.0]*len(line_list)
)

# Plotting
plot_dir = "./plots"
os.makedirs(plot_dir, exist_ok=True)

x = normalized_spec.spectral_axis.to(u.AA).value
y = normalized_spec.flux.value
y_fit = best_model(x)

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 4), sharey=True)
ax1.plot(x, y, lw=1)
ax1.set(title='Normalized Spectrum', xlabel='Wavelength (Å)', ylabel='Flux')

ax2.plot(x, y, label='Data', lw=1)
ax2.plot(x, y_fit, label='Seric Model', lw=1.5)
ax2.set(title='Multi-Seric Fit', xlabel='Wavelength (Å)')
ax2.legend()

ax3.plot(x, y - y_fit, lw=1)
ax3.axhline(0, linestyle='--')
ax3.set(title='Residuals', xlabel='Wavelength (Å)')

plt.tight_layout()
plt.savefig(os.path.join(plot_dir, 'multi_seric_fit.pdf'))
plt.show()

# Display results
print(results)
```
